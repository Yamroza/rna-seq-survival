```python
import gseapy as gp
import json
import pandas as pd
```

```python
def load_gene_runs(filepath):
    runs = []
    with open(filepath) as f:
        for line in f:
            runs.append(json.loads(line))
    return runs

runs = load_gene_runs("../analysis/training_shap.jsonl")
```

```python
try:
    gene_sets = gp.get_library(name='DisGeNET', organism='Human')
    print(f"DisGeNET library fetched successfully.")
except Exception as e:
    print(f"API Error: {e}")
    gene_sets = {}

search_configs = {
    "Breast Cancer": [
        "Breast Cancer", "Breast Neoplasm", "Carcinoma of breast",
        "Ductal Carcinoma", "Invasive Breast Carcinoma"
    ],
}
```

```python
DISEASE_GENES_POOL = {}

for disease_name, keywords in search_configs.items():
    collected_genes = set()
    matched_terms = []

    for db_term, genes in gene_sets.items():
        if any(key.lower() in db_term.lower() for key in keywords):
            collected_genes.update(genes)
            matched_terms.append(db_term)

        DISEASE_GENES_POOL[disease_name] = collected_genes

class GeneDataset:
    def __init__(self, gene_pool):
        self.disease_genes = gene_pool

dataset = GeneDataset(DISEASE_GENES_POOL)
```

```python
# top 1000 genes from each fold
gene_lists = [run["genes"][:1000] for run in runs]
top_concat = [element for gene_list in gene_lists for element in gene_list]

gene_sets = {run["run_name"]: set(run["genes"]) for run in runs}
common_all = set.intersection(*gene_sets.values())

top_concat.extend(common_all)
len(set(top_concat))
```
