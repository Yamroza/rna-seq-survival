```python
import json
from pathlib import Path
import pandas as pd
```

```python
results_dir = Path("/scratch/2370352/my-research/repo/src/results/dss_survival_brca/SNN_big")  # <- zmień jeśli trzeba
save_directory = '../data/results_gathered/results_gathered_snn_big.csv'
```

```python
rows = []

for json_file in results_dir.glob("*.json"):
    if json_file.name == "args.json":
        continue
    
    with open(json_file) as f:
        data = json.load(f)
    try:
        summary = data["Summary"]
    except:
        continue
    
    row = {
        "experiment": json_file.stem,  # nazwa pliku bez .json
        "loss_avg": summary["loss"]["avg"],
        "loss_std": summary["loss"]["std"],
        "c_index_avg": summary["c_index"]["avg"],
        "c_index_std": summary["c_index"]["std"],
        "c_index_ipcw_avg": summary["c_index_ipcw"]["avg"],
        "c_index_ipcw_std": summary["c_index_ipcw"]["std"],
    }

    rows.append(row)

df = pd.DataFrame(rows)
df
```

```python
df = df.sort_values(by=["c_index_avg"], ascending=False)
```

```python
df.to_csv(save_directory)
```
