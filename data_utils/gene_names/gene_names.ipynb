```python
import pandas as pd
import anndata as ad
```

wybieramy geny w zapisie census, czyli adata

```python
data_path = '../../data/0_data_for_mlp/TCGA-BRCA.star_tpm.csv'
df_tcga = pd.read_csv(data_path)
tcga_genes = df_tcga.columns[1:]
```

```python
adata = ad.read_h5ad("../../adapter_premium/data_new/blkb_common_path_test.h5ad")
adata_genes = adata.var['feature_name'].tolist()
```

```python
overlap = set(adata_genes) & set(tcga_genes)
missing = set(tcga_genes) - overlap

print(f'len adata_genes: {len(adata_genes)}')
print(f'len tcga_genes: {len(tcga_genes)}')
print(f'len overlap: {len(overlap)}')
print(f'len missing: {len(missing)}')
```

```python
gene_info = pd.read_csv("../../data/gene_info_table.csv")

# Stwórz słownik: gene_name -> ensembl_id
symbol_to_ensg_table = dict(zip(gene_info['gene_name'], gene_info['ensembl_id']))
ensg_to_symbol_table = dict(zip(gene_info['ensembl_id'], gene_info['gene_name']))

# jeśli w adata masz geny zapisane jako ENSG, mapujemy je
mapped_genes = [symbol_to_ensg_table[g] for g in missing if g in symbol_to_ensg_table]
```

```python
overlap_subset = set(adata_genes) & set(mapped_genes)
overlap_subset_tcga = [ensg_to_symbol_table[g] for g in overlap_subset]

overlap.update(overlap_subset)

missing -= set(overlap_subset_tcga)
print(f'len overlap: {len(overlap_subset)}')
print(f'len missing: {len(missing)}')
```

```python
gene_info = pd.read_csv("../../data/gene_info.csv")

# Stwórz słownik: gene_name -> ensembl_id
symbol_to_ensg = dict(zip(gene_info['feature_name'], gene_info['feature_id']))
ensg_to_symbol = dict(zip(gene_info['feature_id'], gene_info['feature_name']))

mapped_genes = [symbol_to_ensg[g] for g in missing if g in symbol_to_ensg]
```

```python
overlap_subset = set(adata_genes) & set(mapped_genes)
overlap_subset_tcga = [ensg_to_symbol[g] for g in overlap_subset]

overlap.update(overlap_subset)

missing -= set(overlap_subset_tcga)
print(f'len overlap: {len(overlap_subset)}')
print(f'len missing: {len(missing)}')
```
