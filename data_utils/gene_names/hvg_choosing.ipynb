```python
import pandas as pd
import scanpy as sc
import anndata as ad
import os
import json
```

```python
data_path = '../../data/0_data_for_mlp_small_cohorts/'
dataset = 'TCGA-BRCA.star_tpm.csv'
n_hvg = 3000

output_path = '../../data/hvg_genes_lists/'

bulk_data_path = os.path.join(data_path, dataset)
output_path = os.path.join(output_path, dataset.replace('.csv', f'_hvg_{n_hvg}.json'))
```

```python

df = pd.read_csv(bulk_data_path, index_col=0)
adata = ad.AnnData(X=df.values)
adata.obs_names = df.index
adata.var_names = df.columns

sc.pp.highly_variable_genes(adata, n_top_genes=n_hvg)
adata = adata[:, adata.var['highly_variable']]
```

```python
genes_list = list(adata.var_names)
```

```python
with open(output_path, 'w') as f:
    json.dump(genes_list, f, indent=2)
```
