```python
import pandas as pd
import glob
import numpy as np
```

```python
folder = "../data/scgpt_embeddings/new_lengths/"

files = glob.glob(folder + "*.csv")

# Wczytaj wszystkie do listy
dfs = [pd.read_csv(f, index_col=0) for f in files]
```

```python

# Stack i mean po osi "ekspertów"
# Wynik: (n_próbek, 512)
combined = pd.concat(dfs, axis=0)  # (20 * n_próbek, 512)

# Grupuj po ID próbki i uśrednij
mean_embeddings = combined.groupby(combined.index).mean()

mean_embeddings.to_csv(f"{folder}/mean_embeddings.csv")
```
