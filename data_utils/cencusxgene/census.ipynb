```python
with cellxgene_census.open_soma(census_version="2025-11-08") as census:

    cell_adata = cellxgene_census.get_anndata(
        census,
        "homo_sapiens",
        obs_value_filter = "sex == 'female' and tissue == 'lung' and cell_type in ['neuron', 'macrophage']",
        column_names = ["assay", "cell_type", "tissue", "tissue_general", "suspension_type", "disease"]
    )
    cell_adata
```

```python
sc.pp.normalize_total(cell_adata, target_sum=1e4, inplace=False)
sc.pp.log1p(cell_adata)
```

```python
with cellxgene_census.open_soma(census_version="2025-11-08") as census:

    cell_metadata_all = cellxgene_census.get_obs(
        census,
        "homo_sapiens",
        value_filter = "tissue == 'breast'",
        # value_filter = "sex == 'female' and cell_type == 'neuron' and tissue == 'lung'",
        # column_names = ["assay", "cell_type", "tissue", "tissue_general", "suspension_type", "disease"]
    )
    cell_metadata_all
```

```python
cell_metadata_all
```

```python
with open(f'{cell_gene_dir}/cell_types_breast.txt', 'w') as f:
    f.write(cell_metadata_all['cell_type'].value_counts().to_string())

with open(f'{cell_gene_dir}/tissues_breast.txt', 'w') as f:
    f.write(cell_metadata_all['tissue'].value_counts().to_string())

with open(f'{cell_gene_dir}/tissues_general_breast.txt', 'w') as f:
    f.write(cell_metadata_all['tissue_general'].value_counts().to_string())
```
