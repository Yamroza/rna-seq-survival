```python
df = pd.read_csv("/scratch/2370352/my-research/repo/src/data/files/tcga_brca/splits/0/train_filtered.csv")
```

```python
import numpy as np
import pandas as pd

times = df['dss_survival_days'].values
cens = df['dss_censorship'].values  # 0=event, 1=censored

# Funkcja censoringu w surowym ujęciu
df = pd.DataFrame({'dss_survival_days': times, 'dss_censorship': cens})
df = df.sort_values('dss_survival_days')

# Sprawdzenie, czy w którymś czasie wszyscy są ocenzurowani
for t in np.unique(times):
    subset = df[df['dss_survival_days'] == t]
    if subset['dss_censorship'].sum() == len(subset):
        print(f"Problem: wszystkie obserwacje w czasie {t} są ocenzurowane")
```
