```python
import os
import json
import pandas as pd
```

```python

input_dir = "/scratch/2370352/my-research/ts_survival_prediction/src/results/dss_survival_brca/mlp_ica"   # folder z JSON-ami
output_file = "summary_results_mlp_ica.csv"

rows = []

for fname in os.listdir(input_dir):
    if fname.endswith(".json"):
        path = os.path.join(input_dir, fname)

        with open(path, "r") as f:
            data = json.load(f)

        summary = data.get("Summary", {})

        row = {
            "file": fname,
            "loss_avg": summary.get("loss", {}).get("avg"),
            "loss_std": summary.get("loss", {}).get("std"),
            "c_index_avg": summary.get("c_index", {}).get("avg"),
            "c_index_std": summary.get("c_index", {}).get("std"),
            "c_index_ipcw_avg": summary.get("c_index_ipcw", {}).get("avg"),
            "c_index_ipcw_std": summary.get("c_index_ipcw", {}).get("std"),
        }

        rows.append(row)

# === DataFrame ===
df = pd.DataFrame(rows)

# opcjonalnie: sortowanie po najlepszym wyniku
df = df.sort_values(by="c_index_ipcw_avg", ascending=False)

# zapis
df.to_csv(output_file, index=False)

print("Zapisano:", output_file)
```
