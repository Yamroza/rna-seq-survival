```python
import numpy as np
import os
import json
```

```python
def get_results(results_dir, data_type, exp_name):
    c_index = []
    c_index_ipcw = []

    for i in range(5):
        file_name = os.path.join(results_dir, f"Fold_{i}/train_test_summary.json")
        if os.path.exists(file_name):
            # open JSON file and store in dict
            with open(file_name, 'r') as f:
                # json
                data = json.load(f)
            c_index.append(data['c_index'])
            c_index_ipcw.append(data['c_index_ipcw'])

    print(f"Results for {data_type} - {exp_name}:")
    if len(c_index_ipcw) < 5:
        print("Not all folds completed.")
    else:
        print(f"c_index: {np.mean(c_index):.3f}±{np.std(c_index):.3f}")
        print(f"c_index_ipcw: {np.mean(c_index_ipcw):.3f}±{np.std(c_index_ipcw):.3f}")
```

```python
# Results after training and then direct testing
for exp_name in ['DIMAF_ST']:#["MLP_RM", "SNN_RM", "MLP", "SNN", "PW_MLP_CONCAT", "PW_SNN_CONCAT","PW_MLP_MEAN", "PW_SNN_MEAN", "DIMAF_SD"]:
    for data_type in ["brca", "blca", "luad", "kirc"]:
        results_dir = f"../results/dss_survival_{data_type}/{exp_name}/"
        try:
            get_results(results_dir, data_type, exp_name)
        except Exception as e:
            print(f"Not yet done with training for {data_type} - {exp_name}")
```

```python
# Results after testing
c_index_test = []
c_index_ipcw_test = []
file_name = os.path.join(results_dir, f"Final_results.json")
if os.path.exists(file_name):
    # open JSON file and store in dict
    with open(file_name, 'r') as f:
        # json
        data = json.load(f)

for key, value in data.items():
    c_index_test.append(value['c_index'])
    c_index_ipcw_test.append(value['c_index_ipcw'])
   

print(f"Final results for {data_type} - {exp_name}:")
print(f"c_index: {np.mean(c_index_test):.3f}±{np.std(c_index_test):.3f}")
print(f"c_index_ipcw: {np.mean(c_index_ipcw_test):.3f}±{np.std(c_index_ipcw_test):.3f}")
```
